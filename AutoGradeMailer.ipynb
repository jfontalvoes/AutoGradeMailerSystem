```python
df = pd.read_excel("/content/drive/MyDrive/Proyects/AutoGradeMailer/StudentGrades.xlsx")
df
```

```python
for i, row in df.iterrows():
  content = {'student_name': row['Student Name'],
             'math_grade': row['Math'],
             'physics_grade': row['Phy'],
             'chemistry_grade': row['Che']
  }
  content.update(constants)

  doc.render(content)
  doc.save(f"Grades of {row['Student Name']}.docx")

  print(content)
```
